```python
import pandas as pd
import requests
import numpy as np
import matplotlib.pyplot as plt
import datetime as dt
```

```python
stats = pd.read_csv('skater_summary.csv')
skaters = pd.read_csv('skaters.csv')
bios = pd.read_csv('nhl_bios.csv')
skaters['defense'] = np.where(skaters['positionCode'] == 'D', 'Yes', 'No')
defense = skaters[['id', 'positionCode', 'defense', 'birthDate']]
defense['birthDate'] = pd.to_datetime(defense['birthDate'])
defense['birthDate'] = defense['birthDate'].dt.year
```

```python
s_merged = pd.merge(defense, stats, left_on='id', right_on= 'skaterId', how='left')
s_merged = pd.merge(s_merged, bios, left_on = 'id', right_on = 'id', how = 'left'  )
s_merged = s_merged.drop(columns = ['skaterId'])
s_merged.info()
```

```python
g_stats = pd.read_csv('goalie_summary.csv')
goalies = pd.read_csv('goalies.csv')
goalies['birthDate'] = pd.to_datetime(goalies['birthDate'])
goalies['birthDate'] = goalies['birthDate'].dt.year
goalies_abr = goalies[['id', 'birthDate']]
g_merged = pd.merge(goalies_abr, g_stats, left_on='id', right_on= 'skaterId', how='left')
g_merged = pd.merge(g_merged, bios, left_on = 'id', right_on = 'id', how = 'left'  )
g_merged = g_merged.drop(columns = ['skaterId'])
g_merged['season'] = g_merged['season'].astype(str)
g_merged.info()
```

```python
s_merged = s_merged.dropna()
g_merged = g_merged.dropna()
```

```python
s_merged['season'] = s_merged['season'].astype('str')
s_merged.info()
```

```python
s_merged['Age'] = s_merged['season'].str[:4].astype(int) - s_merged['birthDate']
g_merged['Age'] = g_merged['season'].str[:4].astype(int) - g_merged['birthDate']
```

```python
mapping = {'Yes' : 1, 'No':0}
s_merged['defense'] = s_merged['defense'].map(mapping)
```

```python
s_merged['fantasyPoints'] = (s_merged['goals']*4) + (s_merged['assists']*3) - (s_merged['pim']*.25) - (s_merged['powerPlayPoints']) + (s_merged['shorthandedPoints']) + (s_merged['shots']*.25) + ((s_merged['defense']*(s_merged['goals'] + s_merged['assists']))*0.5)
```

```python
s_merged[s_merged['season'] == '20252026.0']
```

```python
g_merged['fantasyPoints'] = (g_merged['wins']*1) - (g_merged['goalsAgainst']*1.5) + (g_merged['saves']*.3) + (g_merged['shutouts']*3)
```

```python
centers = s_merged[s_merged['positionCode'] == 'C']
centers = centers[centers['season'] == '20252026.0']
centers['zscore'] = (centers['fantasyPoints'] - centers['fantasyPoints'].mean())/centers['fantasyPoints'].std()
```

```python
lwing = s_merged[s_merged['positionCode'] == 'L']
lwing = lwing[lwing['season'] == '20252026.0']
lwing['zscore'] = (lwing['fantasyPoints'] - lwing['fantasyPoints'].mean())/lwing['fantasyPoints'].std()
```

```python
rwing = s_merged[s_merged['positionCode'] == 'R']
rwing = rwing[rwing['season'] == '20252026.0']
rwing['zscore'] = (rwing['fantasyPoints'] - rwing['fantasyPoints'].mean())/rwing['fantasyPoints'].std()
```

```python
defenders = s_merged[s_merged['positionCode'] == 'D']
defenders = defenders[defenders['season'] == '20252026.0']
defenders['zscore'] = (defenders['fantasyPoints'] - defenders['fantasyPoints'].mean())/defenders['fantasyPoints'].std()
```

```python
goalkeepers = g_merged[g_merged['season'] == '20252026.0']
goalkeepers['zscore'] = (goalkeepers['fantasyPoints'] - goalkeepers['fantasyPoints'].mean())/goalkeepers['fantasyPoints'].std()
```

```python
combined = pd.concat([centers, lwing, rwing, defenders, goalkeepers])
combined['tot_score'] = combined['fantasyPoints']*(combined['zscore'])
combined.sort_values(by = 'tot_score', ascending=False)
```
